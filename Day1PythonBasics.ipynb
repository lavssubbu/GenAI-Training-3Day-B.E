# 7. Even or Odd

```python
number = int(input("Enter Number: "))

if number % 2 == 0:
    print("Even")
else:
    print("Odd")
```

# 8. Grade Calculator

```python
mark = int(input("Enter Mark: "))

if mark >= 90:
    print("A Grade")
elif mark >= 75:
    print("B Grade")
elif mark >= 60:
    print("C Grade")
else:
    print("Fail")
```

# 9. Functions

```python
def welcome():
    print("Welcome to Python Programming!")

welcome()
```

```python
def greet(name):
    print("Welcome", name)

greet("Student")
greet("AI Learner")
```

```python
def add(a, b):
    print("Sum =", a + b)

add(10, 20)
```

```python
def square(number):
    return number * number

result = square(6)

print("Square =", result)
```

```python
def maximum(a, b):
    if a > b:
        return a
    else:
        return b

print("Largest =", maximum(15, 25))
```
